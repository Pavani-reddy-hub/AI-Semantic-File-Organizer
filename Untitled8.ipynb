```python
import matplotlib.pyplot as plt

data['location'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Cities")
plt.show()
```

```python
# Clean the 'rate' column for numerical plotting
data['rate'] = data['rate'].astype(str) # Ensure it's string to use .str accessor
data['rate'] = data['rate'].str.replace('/5', '', regex=False) # Remove '/5'
data['rate'] = pd.to_numeric(data['rate'], errors='coerce') # Convert to numeric, turn errors into NaN

# Drop NaN values from the 'rate' column for plotting
clean_ratings = data.dropna(subset=['rate'])['rate']

# Plot the histogram
plt.hist(clean_ratings)
plt.title("Restaurant Ratings")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.show()
```

```python
import seaborn as sns
import numpy as np

# Clean the 'rate' column again to ensure consistency
data['rate'] = data['rate'].astype(str)
data['rate'] = data['rate'].str.replace('/5', '', regex=False)
data['rate'] = pd.to_numeric(data['rate'], errors='coerce')

# Clean the 'approx_cost(for two people)' column
data['approx_cost(for two people)'] = data['approx_cost(for two people)'].astype(str)
data['approx_cost(for two people)'] = data['approx_cost(for two people)'].str.replace(',', '', regex=False)
data['approx_cost(for two people)'] = pd.to_numeric(data['approx_cost(for two people)'], errors='coerce')

# Drop rows where either 'rate' or 'approx_cost(for two people)' is NaN for plotting
plot_data = data.dropna(subset=['rate', 'approx_cost(for two people)'])

sns.boxplot(x='approx_cost(for two people)',
            y='rate',
            data=plot_data)

plt.title("Restaurant Rating by Approximate Cost")
plt.xlabel("Approximate Cost for Two People")
plt.ylabel("Rating")
plt.show()
```

```python
sns.heatmap(
    data.corr(numeric_only=True),
    annot=True
)

plt.show()
```
